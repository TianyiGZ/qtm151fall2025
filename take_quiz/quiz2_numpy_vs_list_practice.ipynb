## Q1 — List concatenation vs elementwise addition
Intended: `[1,3,5] + [5,5,5] → [6,8,10]` (elementwise).

## Q2 — List repetition vs scalar multiplication
Intended: multiply each element by 2 → `[2,4,6]`. Lists will *repeat*, not multiply.

```python
# --- Practice ---
lst = [1, 2, 3]
x = lst * 2
print(x)  # <-- Your comment: list * 2 repeats the list, not elementwise multiply
```

```python
# --- Solution ---
arr = np.array([1, 2, 3])
print(arr * 2)
```

## Q3 — List + scalar vs array + scalar (broadcasting)
Intended: add 1 to each element → `[2,3,4]`.

```python
# --- Practice ---
lst = [1, 2, 3]
try:
    print(lst + 1)  # TypeError in plain Python
except TypeError as e:
    print('TypeError:', e, '\n', 'Your comment: lists do not support + scalar; arrays do via broadcasting')
```

```python
# --- Solution ---
arr = np.array([1, 2, 3])
print(arr + 1)
```

## Q4 — Elementwise multiply two sequences
Intended: `[1,2,3] * [4,5,6] → [4,10,18]` (elementwise).

```python
# --- Practice ---
a = [1, 2, 3]
b = [4, 5, 6]
try:
    print(a * b)  # TypeError: list * list is invalid; only repeat by integer
except TypeError as e:
    print('TypeError:', e, '\n', 'Your comment: list * list is not elementwise multiply')
```

```python
# --- Solution ---
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a * b)
```

## Q5 — Mixed types inside a list
Intended: treat numeric-looking strings as numbers before math.

```python
# --- Practice ---
lst = [1, '2', 3]
try:
    print(np.array(lst) + 1)  # will upcast to <U (string) and not compute as numbers
except Exception as e:
    print('Unexpected:', e)
print(np.array(lst).dtype, '← Your comment: dtype becomes string, so math is wrong/blocked')
```

```python
# --- Solution ---
lst = [1, '2', 3]
arr = np.array(list(map(float, lst)))
print(arr + 1)
```

## Q6 — Broadcasting with shapes
Intended: add a column vector to a row vector to form a matrix.

```python
# --- Practice ---
a = np.array([1, 2, 3])      # shape (3,)
b = np.array([[10], [20]])   # shape (2,1)
try:
    print(a + b)  # Works via broadcasting to (2,3)
except ValueError as e:
    print('ValueError:', e)
print('Shapes:', a.shape, b.shape)
```

```python
# --- Solution ---
a = np.array([1, 2, 3]).reshape(1, 3)  # (1,3)
b = np.array([[10], [20]])             # (2,1)
print(a + b)                           # result is (2,3)
```

## Q7 — Shape mismatch (classic broadcasting error)
Intended: elementwise add arrays of **same length**.

```python
# --- Practice ---
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30])
try:
    print(a + b)
except ValueError as e:
    print('ValueError:', e, '\nYour comment: lengths differ; cannot broadcast (4,) with (3,)')
```

```python
# --- Solution ---
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])
print(a + b)
```
