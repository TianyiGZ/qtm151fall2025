## Q1 — TypeError (incompatible types)
Fix returning `'72 inches'` by resolving a type mismatch.

```python
# --- Practice ---
def as_inches(x):
    return x + ' inches'
print(as_inches(72))
```

```python
# --- Solution ---
def as_inches(x):
    return str(x) + ' inches'
print(as_inches(72))
```

## Q2 — ValueError (right type, bad value)
Convert the string to a number safely.

```python
# --- Practice ---
s = '3.14'
n = int(s)
n
```

```python
# --- Solution ---
s = '3.14'
n = float(s)
n_int = int(float(s))

print(n, n_int)
```

## Q5 — KeyError (missing key/column)

```python
# --- Practice ---
prefs['flavor']
people['height'].mean()
```

## Q7 — ZeroDivisionError
Guard against zero denominator.

```python
# --- Practice ---
hits, total = 5, 0
hits / total
```

```python
# --- Solution ---
hits, total = 5, 0
hits / total if total else float('nan')
```
