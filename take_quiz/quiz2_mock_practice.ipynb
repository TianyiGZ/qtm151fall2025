```python
# --- Solution ---
def bmi_flag(h_in, w_lb):
    h_m = h_in * 0.0254
    w_kg = w_lb * 0.453592
    bmi = w_kg / (h_m ** 2)
    return 'Overweight' if bmi >= 25 else 'Normal'

df['BMI_Flag'] = df.apply(lambda r: bmi_flag(r['Height_Inches'], r['Weight_Pounds']), axis=1)
print(df['BMI_Flag'].value_counts())
```

## Q6. Side-by-side scatter plots
Plot Height vs. Weight for Group A (left) and Group B (right) in two subplots, sharing axes.

```python
# --- Practice ---
# TODO: create side-by-side scatter plots for A and B
```

```python
# --- Solution ---
group_A = df.query("Group == 'A'")
group_B = df.query("Group == 'B'")

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharex=True, sharey=True)
axes[0].scatter(group_A['Height_Inches'], group_A['Weight_Pounds'])
axes[0].set_title('Group A')
axes[0].set_xlabel('Height (in)')
axes[0].set_ylabel('Weight (lb)')

axes[1].scatter(group_B['Height_Inches'], group_B['Weight_Pounds'])
axes[1].set_title('Group B')
axes[1].set_xlabel('Height (in)')

plt.tight_layout()
plt.show()
```
